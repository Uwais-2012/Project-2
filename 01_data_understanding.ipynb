```python

# Import pandas to work with tabular data
import pandas as pd

# Load the food delivery CSV file
df = pd.read_csv("data/ONINE_FOOD_DELIVERY_ANALYSIS.csv")

# Display the first 5 rows
df.head()
```

```python

# Display the number of rows and columns
print("Dataset shape:", df.shape)

# Display the column names
print("\nColumn names:")
print(df.columns.tolist())

# Check the data types
print("\nData types:")
print(df.dtypes)
```

```python

# Count missing values in each column
df.isnull().sum()
```

```python

# Calculate missing values as percentages
null_percent = df.isnull().mean() * 100

# Display percentages from highest to lowest
print(null_percent.sort_values(ascending=False))
```

```python

print(df.dtypes)
```

```python

# Check the mean of customer age
print("Mean age:", df["Customer_Age"].mean())

# Check the median of customer age
print("Median age:", df["Customer_Age"].median())

# Check the mode of customer gender
print("Mode of customer gender:", df["Customer_Gender"].mode())
```

```python

# Create a copy of the original dataset
df_clean = df.copy()

# Fill missing customer ages with the median
df_clean["Customer_Age"] = df_clean["Customer_Age"].fillna(
    df_clean["Customer_Age"].median()
)

# Check the remaining missing customer ages
print("Missing ages before cleaning:", df["Customer_Age"].isnull().sum())
print("Missing ages after cleaning:", df_clean["Customer_Age"].isnull().sum())
```

```python
# Check the number of missing ages after cleaning
print("Missing ages:", df_clean["Customer_Age"].isnull().sum())

# Display the median age after cleaning
print("Median age:", df_clean["Customer_Age"].median())

# Display the first 5 customer ages
df_clean["Customer_Age"].head()
```

```python

# Display statistical information about delivery time
print(df["Delivery_Time_Min"].describe())

# Count missing delivery times
print("\nMissing delivery times:")
print(df["Delivery_Time_Min"].isnull().sum())

# Calculate the percentage of missing delivery times
print("\nMissing percentage:")
print(df["Delivery_Time_Min"].isnull().mean() * 100)
```

```python
# Count the number of orders in each status
print(df["Order_Status"].value_counts())

# Count missing delivery times for each order status
print(df.groupby("Order_Status")["Delivery_Time_Min"].apply(
    lambda x: x.isnull().sum()
))
```

```python
# Find the median delivery time for delivered orders with recorded times
delivered_median = df_clean.loc[
    (df_clean["Order_Status"] == "Delivered") &
    (df_clean["Delivery_Time_Min"].notnull()),
    "Delivery_Time_Min"
].median()

# Fill missing delivery times for delivered orders only
df_clean.loc[
    (df_clean["Order_Status"] == "Delivered") &
    (df_clean["Delivery_Time_Min"].isnull()),
    "Delivery_Time_Min"
] = delivered_median

# Display the median used for filling missing values
print("Median delivery time:", delivered_median)

# Check missing delivery times for each order status
print("\nMissing delivery times after cleaning:")
print(
    df_clean.groupby("Order_Status")["Delivery_Time_Min"].apply(
        lambda x: x.isnull().sum()
    )
)
```

```python
# Check the number of missing delivery times
print("Missing delivery times:", df_clean["Delivery_Time_Min"].isnull().sum())

# Check the delivery-time statistics after cleaning
print("\nDelivery time statistics:")
print(df_clean["Delivery_Time_Min"].describe())
```

```python
# Display statistical information about order value
print(df["Order_Value"].describe())

# Count missing order values
print("\nMissing order values:")
print(df["Order_Value"].isnull().sum())

# Calculate the percentage of missing order values
print("\nMissing percentage:")
print(df["Order_Value"].isnull().mean() * 100)
```

```python
# Count orders by order status
print(df["Order_Status"].value_counts())

# Count missing order values for each order status
print(
    df.groupby("Order_Status")["Order_Value"].apply(
        lambda x: x.isnull().sum()
    )
)
```

```python
# Display statistics for final amount
print(df["Final_Amount"].describe())

# Count missing final amounts
print("\nMissing final amounts:")
print(df["Final_Amount"].isnull().sum())

# Check order value and final amount together
print("\nOrder value and final amount:")
print(df[["Order_Value", "Discount_Applied", "Final_Amount"]].head(10))
```

```python
# Display rows with negative final amounts
print(df[df["Final_Amount"] < 0][
    ["Order_ID", "Order_Status", "Order_Value",
     "Discount_Applied", "Final_Amount"]
])

# Count negative final amounts
print("\nNumber of negative final amounts:")
print((df["Final_Amount"] < 0).sum())
```

```python
# Count the different discount values
print(df["Discount_Applied"].value_counts(dropna=False).sort_index())

# Check how many orders have a discount greater than their order value
print("\nDiscount greater than order value:")
print(
    (
        df["Discount_Applied"].notnull()
        & df["Order_Value"].notnull()
        & (df["Discount_Applied"] > df["Order_Value"])
    ).sum()
)
```

```python
# Select rows where all three amount columns have values
amount_check = df[
    df["Order_Value"].notnull()
    & df["Discount_Applied"].notnull()
    & df["Final_Amount"].notnull()
].copy()

# Calculate the expected final amount
amount_check["Expected_Final_Amount"] = (
    amount_check["Order_Value"] - amount_check["Discount_Applied"]
)

# Count rows where the actual and expected amounts differ
print("Rows where final amount does not match the formula:")
print(
    (
        amount_check["Final_Amount"]
        != amount_check["Expected_Final_Amount"]
    ).sum()
)

# Display a few rows where the amounts differ
print(
    amount_check[
        amount_check["Final_Amount"]
        != amount_check["Expected_Final_Amount"]
    ][
        ["Order_Value", "Discount_Applied", "Final_Amount",
         "Expected_Final_Amount"]
    ].head(10)
)
```

```python
# Count rows where order value is missing but final amount and discount are available
can_calculate = (
    df["Order_Value"].isnull()
    & df["Final_Amount"].notnull()
    & df["Discount_Applied"].notnull()
)

print("Missing order values that can be calculated:")
print(can_calculate.sum())

# Calculate a few missing order values using the formula
print(
    df.loc[
        can_calculate,
        ["Order_Value", "Discount_Applied", "Final_Amount"]
    ].head(10)
)
```

```python
# Check missing values in the three amount columns
print("Missing Order_Value:", df["Order_Value"].isnull().sum())
print("Missing Discount_Applied:", df["Discount_Applied"].isnull().sum())
print("Missing Final_Amount:", df["Final_Amount"].isnull().sum())

# Count rows where all three amount columns are missing
all_three_missing = (
    df["Order_Value"].isnull()
    & df["Discount_Applied"].isnull()
    & df["Final_Amount"].isnull()
)

print("\nRows where all three amounts are missing:")
print(all_three_missing.sum())

# Count rows where order value is missing but another amount is available
order_value_missing = (
    df["Order_Value"].isnull()
    & (
        df["Discount_Applied"].notnull()
        | df["Final_Amount"].notnull()
    )
)

print("\nMissing order values with another amount available:")
print(order_value_missing.sum())
```

```python
# Check missing amounts for each order status
for column in ["Order_Value", "Discount_Applied", "Final_Amount"]:

    print(f"\nMissing {column} by Order_Status:")

    print(
        df.groupby("Order_Status")[column]
        .apply(lambda x: x.isnull().sum())
    )
```

```python

# Check the total number of orders in each status
print("Total orders by status:")
print(df["Order_Status"].value_counts())

# Calculate missing-value percentages for each amount column and status
for column in ["Order_Value", "Discount_Applied", "Final_Amount"]:
    print(f"\nMissing percentage for {column}:")

    missing_percentage = (
        df.groupby("Order_Status")[column]
        .apply(lambda x: x.isnull().mean() * 100)
    )

    print(missing_percentage.round(2))
```

```python

# Find rows where final amount is missing but the other two amounts are available
can_calculate_final = (
    df["Final_Amount"].isnull()
    & df["Order_Value"].notnull()
    & df["Discount_Applied"].notnull()
)

# Count how many final amounts can be calculated
print("Final amounts that can be calculated:")
print(can_calculate_final.sum())

# Display a few examples
print(
    df.loc[
        can_calculate_final,
        ["Order_Value", "Discount_Applied", "Final_Amount"]
    ].head(10)
)
```

```python

# Find missing final amounts where order value and discount are available
can_calculate_final = (
    df_clean["Final_Amount"].isnull()
    & df_clean["Order_Value"].notnull()
    & df_clean["Discount_Applied"].notnull()
)

# Calculate the missing final amounts using the verified formula
df_clean.loc[can_calculate_final, "Final_Amount"] = (
    df_clean.loc[can_calculate_final, "Order_Value"]
    - df_clean.loc[can_calculate_final, "Discount_Applied"]
)

# Check how many final amounts are still missing
print("Final amounts filled:", can_calculate_final.sum())
print("Final amounts still missing:", df_clean["Final_Amount"].isnull().sum())
```

```python

# Check which amount information is available for remaining missing final amounts
remaining_final = df_clean["Final_Amount"].isnull()

print("Remaining missing final amounts:")
print(remaining_final.sum())

print("\nOrder value available:")
print(df_clean.loc[remaining_final, "Order_Value"].notnull().sum())

print("\nDiscount available:")
print(df_clean.loc[remaining_final, "Discount_Applied"].notnull().sum())

print("\nBoth order value and discount missing:")
print(
    (
        df_clean.loc[remaining_final, "Order_Value"].isnull()
        & df_clean.loc[remaining_final, "Discount_Applied"].isnull()
    ).sum()
)


```

```python

# Count rows where final amount is missing and both inputs are available
can_calculate_final = (
    df_clean["Final_Amount"].isnull()
    & df_clean["Order_Value"].notnull()
    & df_clean["Discount_Applied"].notnull()
)

print("Additional final amounts that can be calculated:")
print(can_calculate_final.sum())
```

```python

# Find rows where discount is missing
missing_discount = df_clean["Discount_Applied"].isnull()

# Count missing discounts with an available order value
print("Missing discount with order value available:")
print(df_clean.loc[missing_discount, "Order_Value"].notnull().sum())

# Count missing discounts with an available final amount
print("\nMissing discount with final amount available:")
print(df_clean.loc[missing_discount, "Final_Amount"].notnull().sum())

# Count missing discounts where both other amounts are available
print("\nMissing discount with both amounts available:")
print(
    (
        df_clean.loc[missing_discount, "Order_Value"].notnull()
        & df_clean.loc[missing_discount, "Final_Amount"].notnull()
    ).sum()
)
```

```python

# Check discount values and their counts
print(df_clean["Discount_Applied"].value_counts(dropna=False).sort_index())

# Calculate the percentage of missing discounts
missing_discount_percentage = (
    df_clean["Discount_Applied"].isnull().mean() * 100
)

print("\nMissing discount percentage:")
print(round(missing_discount_percentage, 2), "%")
```
